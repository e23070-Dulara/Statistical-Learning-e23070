## Part B

Develop a python based method to filter a sequence of noisy GPS obtained position meaurements.

```python
# ==========================================================
# Kalman Filter for Noisy GPS Position Measurements
# ==========================================================

import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Simulation Parameters
# -----------------------------
dt = 1.0                  # Time interval (seconds)
num_steps = 100           # Number of measurements

# -----------------------------
# State Transition Matrix (A)
# State = [px, py, vx, vy]
# -----------------------------
A = np.array([
    [1, 0, dt, 0],
    [0, 1, 0, dt],
    [0, 0, 1, 0],
    [0, 0, 0, 1]
])

# -----------------------------
# Measurement Matrix (H)
# GPS measures only position
# -----------------------------
H = np.array([
    [1, 0, 0, 0],
    [0, 1, 0, 0]
])

# -----------------------------
# Process Noise Matrix (G)
# -----------------------------
G = np.array([
    [0.5 * dt**2, 0],
    [0, 0.5 * dt**2],
    [dt, 0],
    [0, dt]
])

# Process noise covariance
Sigma_p = np.array([
    [0.2, 0],
    [0, 0.2]
])

Q = G @ Sigma_p @ G.T

# Measurement noise covariance
R = np.array([
    [9, 0],
    [0, 9]
])

# -----------------------------
# Initial State
# -----------------------------
x_est = np.array([
    [0],
    [0],
    [1],
    [0.5]
])

P = np.eye(4) * 10

# -----------------------------
# Generate True Trajectory
# -----------------------------
true_states = []
measurements = []

x_true = x_est.copy()

for i in range(num_steps):

    process_noise = G @ np.random.multivariate_normal(
        [0, 0], Sigma_p
    ).reshape(2,1)

    x_true = A @ x_true + process_noise

    measurement_noise = np.random.multivariate_normal(
        [0,0], R
    ).reshape(2,1)

    y = H @ x_true + measurement_noise

    true_states.append(x_true.flatten())
    measurements.append(y.flatten())

true_states = np.array(true_states)
measurements = np.array(measurements)

# -----------------------------
# Kalman Filter
# -----------------------------
filtered_states = []

for z in measurements:

    z = z.reshape(2,1)

    # Prediction Step
    x_pred = A @ x_est
    P_pred = A @ P @ A.T + Q

    # Innovation
    y = z - H @ x_pred

    # Innovation covariance
    S = H @ P_pred @ H.T + R

    # Kalman Gain
    K = P_pred @ H.T @ np.linalg.inv(S)

    # Update Step
    x_est = x_pred + K @ y
    P = (np.eye(4) - K @ H) @ P_pred

    filtered_states.append(x_est.flatten())

filtered_states = np.array(filtered_states)

# -----------------------------
# Plot Results
# -----------------------------
plt.figure(figsize=(8,6))

plt.plot(true_states[:,0], true_states[:,1],
         label="True Path", linewidth=2)

plt.scatter(measurements[:,0], measurements[:,1],
            color='red', s=20, alpha=0.5,
            label="Noisy GPS")

plt.plot(filtered_states[:,0], filtered_states[:,1],
         color='green', linewidth=2,
         label="Kalman Filter")

plt.xlabel("X Position")
plt.ylabel("Y Position")
plt.title("GPS Position Filtering using Kalman Filter")
plt.legend()
plt.grid(True)

plt.show()
```
